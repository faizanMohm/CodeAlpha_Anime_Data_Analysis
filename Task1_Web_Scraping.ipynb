```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# Humara data is list mein save hoga
anime_data = []

# Headers dalna zaroori hai taaki website ko lage ki ek real browser se request aa rahi hai
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

print("Scraping start kar rahe hain bhai... Thoda wait karna (Lagbhag 1-2 minute lagte hain 1000 data me)\n")

# MyAnimeList me har page par 50 anime hote hain. 1000 ke liye hum limit 0 se 950 tak chalayenge
for i in range(0, 1000, 50):
    url = f"https://myanimelist.net/topanime.php?limit={i}"
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        print(f"Fetching Data from: {url}")
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # Har anime ka data 'tr' tag me class 'ranking-list' ke andar hota hai
        anime_rows = soup.find_all('tr', class_='ranking-list')
        
        for row in anime_rows:
            try:
                # Rank nikalna
                rank = row.find('td', class_='rank ac').find('span').text.strip()
                
                # Title nikalna
                title = row.find('div', class_='di-ib clearfix').find('h3').find('a').text.strip()
                
                # Rating/Score nikalna
                score = row.find('td', class_='score ac fs14').find('span').text.strip()
                
                # Information (Episodes aur Release Dates) nikalna
                info = row.find('div', class_='information di-ib mt4').text.strip().split('\n')
                episodes_type = info[0].strip() # Jaise: TV (24 eps)
                release_date = info[1].strip()  # Jaise: Apr 2009 - Jul 2010
                
                # Data ko dictionary me store karna
                anime_data.append({
                    'Rank': rank,
                    'Title': title,
                    'Score': score,
                    'Format & Episodes': episodes_type,
                    'Release Date': release_date
                })
            except AttributeError:
                # Agar kisi row me data missing ho toh aage badh jao
                continue
    else:
        print(f"Error aayi page {url} pe, Status Code: {response.status_code}")
    
    # Har page scrape karne ke baad 2 second ruko taaki MyAnimeList block na kare
    time.sleep(2)

# Ab is list ko Pandas DataFrame me convert kar do
df = pd.DataFrame(anime_data)

# Data ko CSV me save kar do
file_name = 'Top_1000_Anime_CodeAlpha.csv'
df.to_csv(file_name, index=False)

print(f"\nBhai Task 1 Complete! Data successfully '{file_name}' naam ki file me save ho gaya hai.")
print(f"Total anime scraped: {len(df)}")
```
