```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. DATA LOAD KARNA
print("Step 1: Data Load ho raha hai...")
df = pd.read_csv('Top_1000_Anime_CodeAlpha.csv')

# 2. DATA KO SAMAJHNA AUR CLEAN KARNA
print("Step 2: Data ki safai (Cleaning) chal rahi hai...")

# 'Score' wale column ko text se Number (Float) mein badalna zaroori hai
df['Score'] = pd.to_numeric(df['Score'], errors='coerce')

# Agar kisi anime ka score missing hai (NaN), toh us row ko hata do
df.dropna(subset=['Score'], inplace=True)

# 'Format & Episodes' (Jaise "TV (24 eps)") me se sirf pehla word (Jaise "TV") nikalna
df['Main Format'] = df['Format & Episodes'].apply(lambda x: str(x).split()[0])

print("Data ekdum saaf ho gaya hai! Ab graphs ban rahe hain...\n")

# 3. DATA VISUALIZATION (Graph Banana aur Save karna)

# Graph 1: Sabse zyada kaunsa Format banta hai? (TV, Movie, OVA)
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Main Format', order=df['Main Format'].value_counts().index, palette='magma')
plt.title("Top 1000 Anime me Formats ka Trend")
plt.xlabel("Anime Format")
plt.ylabel("Total Anime Count")

# Graph ko pehle SAVE karenge (.png image me)
plt.savefig('Format_Trend_Graph.png', bbox_inches='tight')
print("-> Pehla graph 'Format_Trend_Graph.png' naam se download/save ho gaya hai.")

# Phir Graph ko screen par SHOW karenge
plt.show() 
# (Dhyan rakhna: Jab ye graph khulega, toh isko (X) se close karna, tabhi dusra graph aayega)


# Graph 2: Ratings (Score) kitni milti hai zyada tar anime ko?
plt.figure(figsize=(10, 5))
sns.histplot(df['Score'], bins=20, kde=True, color='cyan', edgecolor='black')
plt.title("Anime Ratings (Scores) ka Pattern")
plt.xlabel("Rating (Score)")
plt.ylabel("Kitne Anime ko ye rating mili")

# Graph ko save karna
plt.savefig('Score_Pattern_Graph.png', bbox_inches='tight')
print("-> Dusra graph 'Score_Pattern_Graph.png' naam se download/save ho gaya hai.")

# Phir show karna
plt.show()

print("\nBhai Task 2 Complete! Tumhare folder me dono graphs save ho chuke hain.")
```
